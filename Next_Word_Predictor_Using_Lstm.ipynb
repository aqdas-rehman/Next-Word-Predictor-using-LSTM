# **Tokenizer**
We first used tokenizer to give every word an index and to create sequences

```python
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
```

```python
tokenizer = Tokenizer()
```

```python
tokenizer.fit_on_texts([faqs])
```

```python
print(tokenizer.word_index)
```

**We check the total vocabulary in the dataset**

```python
len(tokenizer.word_index)
```

# **Creating sequences using Tokenizer**

```python
for sentence in faqs.split('\n'):
  print(tokenizer.texts_to_sequences([sentence])[0])
```

# **Creating input data**

```python
input_sequences = []
for sentence in faqs.split('\n'):
  tokenized_sentence = tokenizer.texts_to_sequences([sentence])[0]

  for i in range(1,len(tokenized_sentence)):
    input_sequences.append(tokenized_sentence[:i+1])
```

```python
input_sequences
```

```python
max_len = max([len(x) for x in input_sequences])
```

```python
max_len
```

# **Using padding to make them equal**

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
padded_input_sequences = pad_sequences(input_sequences, maxlen = max_len, padding='pre')
```

```python
padded_input_sequences
```

# **Creating X AND Y**

```python
X = padded_input_sequences[:,:-1]
```

```python
y = padded_input_sequences[:,-1]
```

```python
X.shape
```

```python
y.shape
```

# **Using OHE On Y**

```python
from tensorflow.keras.utils import to_categorical
y = to_categorical(y,num_classes=455)
```

```python
y.shape
```

```python
y[1]
```

# ***`Creating Lstm Model`***

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense
```

```python
model = Sequential()
model.add(Embedding(455, 100, input_length=41))
model.add(LSTM(150))
model.add(Dense(455, activation='softmax'))
```

```python
model.compile(loss='categorical_crossentropy', optimizer='adam',metrics=['accuracy'])
```

```python
model.fit(X,y,epochs=50)
```

# ***`Testing the model`***

```python
import numpy as np
```

```python
text = "Aqdas is a Data Science graduate"


for i in range(10):
  # tokenize
  token_text = tokenizer.texts_to_sequences([text])[0]
  # padding
  padded_token_text = pad_sequences([token_text], maxlen=56, padding='pre')
  # predict
  pos = np.argmax(model.predict(padded_token_text))

  for word,index in tokenizer.word_index.items():
    if index == pos:
      text = text + " " + word
      print(text)
```
